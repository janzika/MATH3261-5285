### Inertial Oscillations

As we discussed in class and saw in Chapter 2 of CRB, in a rotating refence frame, horizontal, unforced motions are described by (eqs. 2.23a and b from CRB)

$$
\frac{d u}{dt} - fv = 0\,\,,\,\,  \frac{d v}{dt} + fu = 0,
$$

where ($u$,$v$) are the zonal and meridional velocities respectively and $f$ is the Coriolis parameter.

From (2.23a and b) it can be inferred that such motions would complete circular motions known as 'intertial oscillations'.

You will be tasked with examining the trajectories of actual floats and assessing whether their is evidence of interial motion.

### Variables, arrays, and indexing

The first thing we need to do is define a numerical grid. We will do this using the `numpy` functions `linspace` and `meshgrid`. Thus, when we call them, we will use the alias we defined for this library, `np`, followed by a period `.` followed by the particular function we want from the library, e.g. `np.linspace`.

First, we create a vector `xv` of `2n+1` equally spaced gridpoints between $x = -1$ and $x = 1$. Again, press **shift + enter** to run the following cell.

````python
# Grid
n = 50
xv = np.linspace(-1,1,2*n+1)

print(xv.shape)
print(xv[0:10])
print(xv[-10:])
````

The command `print(xv.shape)` prints the dimensions of the array `xv`. In this case, it is a 1D array with `2n+1` elements, so the dimensions are $(2n+1,)$. Notice that the second dimension is simply blank, rather than `1`.

The commands `print(xv[0:10])` and `print(xv[-10:])` print the first and last ten elements of the array. Indexing in python is indicated using square brackets `[...]` and is referenced to the first element (for positive indices) or the last element (for negative indices). Thus, `xv[0:10]` can be read as: "first + 0 elements to first + 10 elements", and `xv[-10:]` can be read as: "last - 10 elements to last element".

Next we will define 2D arrays `xx` and `yy` using the function `meshgrid`.

````python
xx,yy = np.meshgrid(xv,xv,indexing='ij')

print(xx.shape)
print(yy.shape)

print(xx[0:10,0:10])
print(yy[0:10,0:10])
````

The array `xx` is an $2n+1 \times 2n+1$ matrix in which each row is the corresponding value of `xv`. Likewise, the array `yy` is an $2n+1 \times 2n+1$ matrix with each column having the corresponding value of `xv`.

### Functions, if and for loops

By creating functions we're able to automate different processes that we can then call on with different input variables.


You can then call the function using `function_name(input_variables)`, in your following code.

It should be noted that Python is sensitive to indentation, after functions or loops there should be a Tab i.e. four spaces.

````python
# Function that adds one your input

def add_one(number):
  output = number + 1  # function code is indented
  return(output)

# call on function by name we've assigned
new_number = add_one(3)

 # print output
print(new_number)
````


````python
# Loading and manipulating 2D data arrays
import pandas as pd

# Define URL
url_float_data = \
'https://raw.githubusercontent.com/janzika/MATH3261-5285/refs/heads/main/data/OSMC_30day_982f_f952_0246_truncated.csv'

# Load data using Pandas
float_data = pd.read_csv(url_float_data,header=0, skiprows=[1]) #Column labels spread over two rows

#Display data
display(float_data)
````

We have loaded the float data into our colab notebook as a [*pandas data frame*](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html). You can think of the data as being in a kind of container with functions associated with it that allow you to do things like averages, plots, subsamples etc. Above we have used the *display* function to show the data.

In the dataframe above we can see all the variables.
When there are more variables it can be useful to get a list of them.

````python
list(float_data)
````

We can manipulate the columns of data frames and add new ones.
Let's change the `time` variable to UTC time and create a numeric time varible that can be used for plotting a time series.

````python
# Convert time column to datetime and ensure UTC
float_data['time'] = pd.to_datetime(float_data['time'])

# Convert time to numeric format for the color map (e.g., seconds or epoch time)
float_data['time_numeric'] = float_data['time'].astype('int64') // 10**9

float_data
````

Above we can see we now have the extra column and `time` is UTC.

We can list the unique float numbers, from the `platform_code`, using `pd.unique()`

````python
# The unique 'platform numbers'
float_data['platform_code'].unique()
````

We can now retrive data for specific by finding all entries in `platform_data` equal to a chosen value.
For example, let's retreive data for float number 1501807.

````python
float_no = 1501807
find_float = float_data['platform_code']==float_no
single_float = float_data[find_float]
single_float
````

Indexing the subset of `float_data` that matches the float number we chose returns a new data frame.

To check we have only got data for one float, we can look at how many unique `platform_code` entries we have.

````python
single_float['platform_code'].unique()
````

### Plotting one float trajectory

Let's make a plot of the location data for `float_no`.

````python
# plot float position
plt.plot(single_float['longitude'],single_float['latitude'],'+')
# define axis labels and title
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Float number '+str(float_no))
````

This float has been drifting around in the Southern Ocean close to the Greenwich Meridian (south of Sotuh Africa maybe?). With some slightly more sophisticated code, we can make a more revealing image with the time shown as a colourscale.

````python
# plot float position and time

plt.scatter(single_float['longitude'],single_float['latitude'],c = single_float['time_numeric'])
# define axis labels and title
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Float number '+str(float_no))
cbar = plt.colorbar()
cbar.set_label('Date')

# Convert ticks back to readable UTC strings on the colorbar
ticks = cbar.get_ticks()
cbar.set_ticklabels(pd.to_datetime(ticks, unit='s').strftime('%D'))

#Put grid lines on background
plt.grid()
````
